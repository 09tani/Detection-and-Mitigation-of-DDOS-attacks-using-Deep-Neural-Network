# Import Libaries including numpy and sklearn

```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report
from sklearn import metrics
```

# Read in the data frame

```python
df = pd.read_csv('../original.csv')
df.head()
```

```python
df.shape
```

## Drop object timestamp and Encode Label

```python
df = df.drop("Timestamp", axis=1)
df["Label"] = df["Label"].map({"Benign": 0, "FTP-BruteForce": 1, "SSH-Bruteforce": 2})
```

## Remove IP 0 - reason explain in paper

```python
df.drop(columns=["Flow Byts/s"], inplace=True)
df = df[df['Protocol'] != 0]
```

## Check for negative vals and remove them

```python
df_cleaned = df[(df >= 0).all(axis=1)]
```

## Check for infinite vals and remove them

```python
inf_check = df_cleaned.isin([np.inf, -np.inf]).any()
columns_with_inf = inf_check[inf_check].index.tolist()
print("Columns with infinities:", columns_with_inf)
```

```python
# Replace inf/-inf with NaN and drop rows
df_cleaned = df_cleaned.replace([np.inf, -np.inf], np.nan).dropna()
print("Remaining infinities in df_cleaned:", np.isinf(df_cleaned).any().any())
```

# Scale the data using standard scaling

```python
scaler = StandardScaler()
df_cleaned_scaled = scaler.fit_transform(df_cleaned)
```

```python
x = df_cleaned.drop(['Label'], axis=1)
y = df_cleaned['Label']
print(y.unique())
```

## Split the data into train and test

- split of 90% train, 10% test

```python
X_train, X_test, y_train, y_test = train_test_split(x, y,test_size = 0.1)
```

```python
X_train.nunique()
```

### Defined accuracy as an array

```python
Classifier_accuracy = []
```

# Building the KNN model using Sklearn package

```python
# 10% of the total data points to be consider x
# prevents over fitting
knn_clf = KNeighborsClassifier(n_neighbors=3000)
knn_clf.fit(X_train, y_train)
y_pred = knn_clf.predict(X_test)
accuracy = metrics.accuracy_score(y_test, y_pred)
Classifier_accuracy.append(accuracy*100)
print("Accuracy of KNN Classifier : %.2f" % (accuracy*100))
```

## SVM

```python
svm_clf = SVC()
svm_clf.fit(X_train, y_train)
y_pred_svm = svm_clf.predict(X_test)
accuracy_svm = metrics.accuracy_score(y_test, y_pred_svm)
Classifier_accuracy.append(accuracy_svm * 100)
print("Accuracy of SVM Classifier : %.2f" % (accuracy_svm * 100))
```

## Gaussian Naive Bayes classifier

```python
nb_clf = GaussianNB()
nb_clf.fit(X_train, y_train)
y_pred_nb = nb_clf.predict(X_test)
accuracy_nb = metrics.accuracy_score(y_test, y_pred_nb)
Classifier_accuracy.append(accuracy_nb * 100)
print("Accuracy of Naive Bayes Classifier : %.2f" % (accuracy_nb * 100))
```
